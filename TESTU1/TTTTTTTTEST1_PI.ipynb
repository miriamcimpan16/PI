1.Let $A = \begin{bmatrix} 4 & 5 \\ -2 & 1 \\ 0 & 6 \end{bmatrix}$ and $B = \begin{bmatrix} 3 & 0 & -1 \\ 2 & 2 & 10 \end{bmatrix}$.Print the shapes of $A$, $B$, and compute $C = B[0] + AB$ , $f = -C[0](x * x)$, and $\nabla_x f$, where $x = A[:, 1]$.

```python
import torch

A=torch.tensor([[4.0,5.0],
                [-2.0,1.0],
                [0.0,6.0]])
B=torch.tensor([[3.0,0.0,-1.0],
                [2.0,2.0,10.0]])
print("Shape of A:", A.shape)
print("Shape of B:", B.shape)
C = B[0] + (A @ B)
print("Matrix C:\n", C)

x = A[:, 1].clone().detach().requires_grad_(True)
f=-torch.dot(C[0],x*x)
print("Value of f:", f)

f.backward()
print("Gradient \u2207_x f:", x.grad)
```

1.The CIFAR100 dataset, which is available in PyTorch, consists of color images of resolution $32 \times 32$. The original training set of $50000$ instances should be split into a training set of $8000$ instances and a validation set of $42000$ instances. Use an MLP with a single hidden layer of $1024$ units and ReLU activation function to classify the CIFAR100 images into $100$ classes. The number of epochs should be $10$, the learning rate of the stochastic gradient descent optimizer should be $0.3$, and the batch size should be $512$. Use dropout after the hidden layer with a dropout rate of $0.2$. Plot the training and validation accuracy. Compute the loss and the accuracy on the original test set.
Test loss 3.99, Test accuracy 9.29

```python
import torch
import torchvision
from torchvision import transforms
from torch import nn
import matplotlib.pyplot as plt

trans = transforms.ToTensor()
cifar_train = torchvision.datasets.CIFAR100(
    root="../data", train=True, transform=trans, download=True
)
cifar_test = torchvision.datasets.CIFAR100(
    root="../data", train=False, transform=trans, download=True
)

cifar_train, cifar_val = torch.utils.data.random_split(
    cifar_train, [8000, 42000], generator=torch.Generator().manual_seed(42)
)

batch_size = 512
train_iter = torch.utils.data.DataLoader(cifar_train, batch_size=batch_size, shuffle=True, num_workers=2)
val_iter = torch.utils.data.DataLoader(cifar_val, batch_size=batch_size, shuffle=False, num_workers=2)
test_iter = torch.utils.data.DataLoader(cifar_test, batch_size=batch_size, shuffle=False, num_workers=2)

net = nn.Sequential(
    nn.Flatten(),
    nn.Linear(3 * 32 * 32, 1024),
    nn.ReLU(),
    nn.Dropout(0.2),
    nn.Linear(1024, 100)
)
loss = nn.CrossEntropyLoss()
lr = 0.3
optimizer = torch.optim.SGD(net.parameters(), lr=lr)

def init_weights(m):
    if type(m) == nn.Linear:
        nn.init.normal_(m.weight, std=0.01)

net.apply(init_weights)

def evaluate_accuracy(net, data_iter):
    net.eval()
    total_loss = 0
    total_hits = 0
    total_samples = 0
    with torch.no_grad():
        for X, y in data_iter:
            y_hat = net(X)
            l = loss(y_hat, y)
            total_loss += float(l)
            total_hits += sum(y_hat.argmax(axis=1).type(y.dtype) == y)
            total_samples += y.numel()
    return float(total_loss) / len(data_iter), float(total_hits) / total_samples * 100

def train_epoch(net, train_iter, loss, optimizer):
    net.train()
    total_loss = 0
    total_hits = 0
    total_samples = 0
    for X, y in train_iter:
        y_hat = net(X)
        l = loss(y_hat, y)
        optimizer.zero_grad()
        l.backward()
        optimizer.step()
        total_loss += float(l)
        total_hits += sum(y_hat.argmax(axis=1).type(y.dtype) == y)
        total_samples += y.numel()
    return float(total_loss) / len(train_iter), float(total_hits) / total_samples * 100

def train(net, train_iter, val_iter, loss, num_epochs, optimizer):
    train_acc_all = []
    val_acc_all = []
    for epoch in range(num_epochs):
        train_loss, train_acc = train_epoch(net, train_iter, loss, optimizer)
        val_loss, val_acc = evaluate_accuracy(net, val_iter)
        train_acc_all.append(train_acc)
        val_acc_all.append(val_acc)
        print(f'Epoch {epoch + 1}, Train loss {train_loss:.2f}, Train accuracy {train_acc:.2f}, Validation loss {val_loss:.2f}, Validation accuracy {val_acc:.2f}')
    return train_acc_all, val_acc_all

num_epochs = 10
train_acc_all, val_acc_all = train(net, train_iter, val_iter, loss, num_epochs, optimizer)

def plot_accuracy(train_acc_all, val_acc_all):
    epochs = range(1, len(train_acc_all) + 1)
    plt.plot(epochs, train_acc_all, 'bo', label='Training acc')
    plt.plot(epochs, val_acc_all, 'b', label='Validation acc')
    plt.title('Training and validation accuracy')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy (%)')
    plt.legend()
    plt.show()

plot_accuracy(train_acc_all, val_acc_all)

test_loss, test_acc = evaluate_accuracy(net, test_iter)
print(f'Test loss {test_loss:.2f}, Test accuracy {test_acc:.2f}')
```

The original training set of $50000$ instances should be split into a training set of $7000$ instances and a validation set of $43000$ instances. Use an MLP with a single hidden layer of $512$ units and ReLU activation function to classify the CIFAR10 images into $10$ classes. test loss 1.91, test accuracy 31.39

```python
import torch
from torchvision import transforms
from torch import nn
import matplotlib.pyplot as plt

trans = transforms.ToTensor()
cifar_train = torchvision.datasets.CIFAR10(
    root="../data", train=True, transform=trans, download=True)
cifar_test = torchvision.datasets.CIFAR10(
    root="../data", train=False, transform=trans, download=True)
cifar_train, cifar_val = torch.utils.data.random_split(cifar_train, [7000, 43000],
                                                           generator=torch.Generator().manual_seed(42))

net = nn.Sequential(nn.Flatten(),
                    nn.Linear(3*32*32, 512),
                    nn.ReLU(),
                    nn.Linear(512, 10))

batch_size=256
train_iter = torch.utils.data.DataLoader(cifar_train, batch_size, shuffle=True, num_workers=2)
val_iter = torch.utils.data.DataLoader(cifar_val, batch_size, shuffle=False, num_workers=2)
test_iter = torch.utils.data.DataLoader(cifar_test, batch_size, shuffle=False, num_workers=2)

def init_weights(m):
    if type(m) == nn.Linear:
        nn.init.normal_(m.weight, std=0.01)

net.apply(init_weights)

lr=0.01
num_epochs = 10
loss = nn.CrossEntropyLoss()
wd = 0.05
optimizer = torch.optim.SGD(net.parameters(), weight_decay=wd, lr=lr)

def evaluate_accuracy(net, data_iter):
    """Compute the accuracy for a model on a dataset."""
    net.eval()  # Set the model to evaluation mode

    total_loss = 0
    total_hits = 0
    total_samples = 0
    with torch.no_grad():
        for X, y in data_iter:
            y_hat = net(X)
            l = loss(y_hat, y)
            total_loss += float(l)
            total_hits += sum(net(X).argmax(axis=1).type(y.dtype) == y)
            total_samples += y.numel()
    return float(total_loss) / len(data_iter), float(total_hits) / total_samples  * 100

def train_epoch(net, train_iter, loss, optimizer):
    # Set the model to training mode
    net.train()
    # Sum of training loss, sum of training correct predictions, no. of examples
    total_loss = 0
    total_hits = 0
    total_samples = 0
    for X, y in train_iter:
        # Compute gradients and update parameters
        y_hat = net(X)
        l = loss(y_hat, y)
        # Using PyTorch built-in optimizer & loss criterion
        optimizer.zero_grad()
        l.backward()
        optimizer.step()
        total_loss += float(l)
        total_hits += sum(y_hat.argmax(axis=1).type(y.dtype) == y)
        total_samples += y.numel()
    # Return training loss and training accuracy
    return float(total_loss) / len(train_iter), float(total_hits) / total_samples  * 100

def train(net, train_iter, val_iter, loss, num_epochs, optimizer):
    """Train a model."""
    train_loss_all = []
    train_acc_all = []
    val_loss_all = []
    val_acc_all = []
    for epoch in range(num_epochs):
        train_loss, train_acc = train_epoch(net, train_iter, loss, optimizer)
        train_loss_all.append(train_loss)
        train_acc_all.append(train_acc)
        val_loss, val_acc = evaluate_accuracy(net, val_iter)
        val_loss_all.append(val_loss)
        val_acc_all.append(val_acc)
        print(f'Epoch {epoch + 1}, Train loss {train_loss:.2f}, Train accuracy {train_acc:.2f}, Validation loss {val_loss:.2f}, Validation accuracy {val_acc:.2f}')

    return train_loss_all, train_acc_all, val_loss_all, val_acc_all

train_loss_all, train_acc_all, val_loss_all, val_acc_all = train(net, train_iter, val_iter, loss, num_epochs, optimizer)

def plot_accuracy(train_acc_all, val_acc_all):
    epochs = range(1, len(train_acc_all) + 1)
    plt.plot(epochs, train_acc_all, 'bo', label='Training acc')
    plt.plot(epochs, val_acc_all, 'b', label='Validation acc')
    plt.title('Training and validation accuracy')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.show()

plot_accuracy(train_acc_all, val_acc_all)

test_loss, test_acc = evaluate_accuracy(net, test_iter)
print(f'Test loss {test_loss:.2f}, Test accuracy {test_acc:.2f}')
```

initialize the weight to 1 and the bias to 0;
implement the corresponding loss function of the ridge regression model and set the hyperparameter lambda to 0.1
;use batch gradient descent and set the learning rate to 0.1; set the number of epochs to 15.  
After the model is trained, compare the learned model parameters with the actual parameters.  
Learned weight tensor([[[1.9503]]])  
True weight tensor([2.])   
Learned bias tensor([2.9433])

```python
import torch
from torch import nn

torch.manual_seed(42)

X = torch.rand(250).reshape(-1, 1)
epsilon = torch.rand(250) / 4
y = 2 * X + 3 + epsilon.reshape(-1, 1)

true_w = torch.tensor([2.])
true_b = 3

net = nn.Sequential(nn.Linear(1, 1))

net[0].weight.data.fill_(1)
net[0].bias.data.fill_(0)

# RIDGE LOSS
lambd = 0.1
mse = nn.MSELoss()

def ridge_loss(y_hat, y):
    return mse(y_hat, y) + lambd * (net[0].weight ** 2).sum()

lr = 0.1
optimizer = torch.optim.SGD(net.parameters(), lr=lr)

num_epochs = 15

for epoch in range(num_epochs):
    l = ridge_loss(net(X), y)

    optimizer.zero_grad()
    l.backward()
    optimizer.step()

    print(f'Epoch {epoch + 1}, Loss {l.item():.4f}')


w = net[0].weight.data
b = net[0].bias.data

print("Learned weight", w)
print("True weight", true_w)
print("Learned bias", b)
print("True bias", true_b)
```

1.The CIFAR10 dataset, which is available in PyTorch, consists of color images of resolution  32×32 . The original training set of  50000  instances should be split into a training set of  3000  instances and a validation set of  47000  instances. Implement the following CNN to classify the CIFAR10 images into  10  classes. The number of epochs should be  5 , the learning rate of the stochastic gradient descent optimizer should be  0.05 , and the batch size should be  256 . The ConvBlock should be implemented in a standalone class, which inherits nn.Module. Conv2d, 32, 1x1 means  32  output channels and  1×1  kernel size. Plot the training and validation accuracy. Print the output shape of the ConvBlock, flatten layer, and output layer. Compute the loss and the accuracy on the original test set. ConvBlock output shape: torch.Size([1, 128, 2, 2])
Flatten output shape: torch.Size([1, 512])
Linear output shape: torch.Size([1, 10])

```python
import torch
import torchvision
from torchvision import transforms
from torch import nn
import matplotlib.pyplot as plt

torch.manual_seed(42)
torch.backends.cudnn.deterministic = True

trans = transforms.ToTensor()
cifar_train = torchvision.datasets.CIFAR100(root="../data", train=True, transform=trans, download=True)
cifar_test = torchvision.datasets.CIFAR100(root="../data", train=False, transform=trans, download=True)
cifar_train, cifar_val = torch.utils.data.random_split(cifar_train, [6000, 44000],generator=torch.Generator().manual_seed(42))

batch_size = 256
train_iter = torch.utils.data.DataLoader(cifar_train, batch_size, shuffle=True, num_workers=2)
val_iter = torch.utils.data.DataLoader(cifar_val, batch_size, shuffle=False, num_workers=2)
test_iter = torch.utils.data.DataLoader(cifar_test, batch_size, shuffle=False, num_workers=2)

class ConvBlock(nn.Module):
    def __init__(self):
        super().__init__()

        # Ramura 1
        self.branch1 = nn.Sequential(
            nn.AvgPool2d(kernel_size=2, stride=16),
            nn.Conv2d(3, 32, kernel_size=1),
            nn.BatchNorm2d(32),
            nn.ReLU()
        )

        # Ramura 2
        self.branch2 = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=1, stride=16),
            nn.BatchNorm2d(32),
            nn.ReLU()
        )

        # Partea comuna pentru ramurile 3 si 4
        self.shared = nn.Sequential(
            nn.Conv2d(3, 64, kernel_size=1),
            nn.ReLU()
        )

        # Ramura 3
        self.branch3 = nn.Sequential(
            nn.Conv2d(64, 32, kernel_size=(1, 3), stride=16),
            nn.ReLU()
        )

        # Ramura 4
        self.branch4 = nn.Sequential(
            nn.Conv2d(64, 32, kernel_size=(3, 1), stride=16),
            nn.ReLU()
        )

    def forward(self, X):
        b1 = self.branch1(X)
        b2 = self.branch2(X)

        Y = self.shared(X)

        b3 = self.branch3(Y)
        b4 = self.branch4(Y)

        return torch.cat((b1, b2, b3, b4), dim=1)
net = nn.Sequential(
    ConvBlock(),
    nn.Flatten(),
    nn.Linear(128 * 2 * 2, 100)
)
X = torch.rand(1, 3, 32, 32)

for layer in net:
    X = layer(X)
    print(layer.__class__.__name__, 'output shape:', X.shape)
def evaluate_accuracy(net, data_iter, loss, device):
    """Compute the accuracy for a model on a dataset."""
    net.eval()  # Set the model to evaluation mode

    total_loss = 0
    total_hits = 0
    total_samples = 0
    with torch.no_grad():
        for X, y in data_iter:
            X, y = X.to(device), y.to(device)
            y_hat = net(X)
            l = loss(y_hat, y)
            total_loss += float(l)
            total_hits += sum(net(X).argmax(axis=1).type(y.dtype) == y)
            total_samples += y.numel()
    return float(total_loss) / len(data_iter), float(total_hits) / total_samples  * 100
def train_epoch(net, train_iter, loss, optimizer, device):
    # Set the model to training mode
    net.train()
    # Sum of training loss, sum of training correct predictions, no. of examples
    total_loss = 0
    total_hits = 0
    total_samples = 0
    for X, y in train_iter:
        # Compute gradients and update parameters
        X, y = X.to(device), y.to(device)
        y_hat = net(X)
        l = loss(y_hat, y)
        # Using PyTorch built-in optimizer & loss criterion
        optimizer.zero_grad()
        l.backward()
        optimizer.step()
        total_loss += float(l)
        total_hits += sum(y_hat.argmax(axis=1).type(y.dtype) == y)
        total_samples += y.numel()
    # Return training loss and training accuracy
    return float(total_loss) / len(train_iter), float(total_hits) / total_samples  * 100
def train(net, train_iter, val_iter, test_iter, num_epochs, lr, device):
    """Train a model."""
    train_loss_all = []
    train_acc_all = []
    val_loss_all = []
    val_acc_all = []
    def init_weights(m):
        if type(m) == nn.Linear or type(m) == nn.Conv2d:
            nn.init.xavier_uniform_(m.weight)
    net.apply(init_weights)
    print('Training on', device)
    net.to(device)
    optimizer = torch.optim.SGD(net.parameters(), lr=lr)
    loss = nn.CrossEntropyLoss()
    for epoch in range(num_epochs):
        train_loss, train_acc = train_epoch(net, train_iter, loss, optimizer, device)
        train_loss_all.append(train_loss)
        train_acc_all.append(train_acc)
        val_loss, val_acc = evaluate_accuracy(net, val_iter, loss, device)
        val_loss_all.append(val_loss)
        val_acc_all.append(val_acc)
        print(f'Epoch {epoch + 1}, Train loss {train_loss:.2f}, Train accuracy {train_acc:.2f}, Validation loss {val_loss:.2f}, Validation accuracy {val_acc:.2f}')
    test_loss, test_acc = evaluate_accuracy(net, test_iter, loss, device)
    print(f'Test loss {test_loss:.2f}, Test accuracy {test_acc:.2f}')

    return train_loss_all, train_acc_all, val_loss_all, val_acc_all
def try_gpu(i=0):
    """Return gpu(i) if exists, otherwise return cpu()."""
    if torch.cuda.device_count() >= i + 1:
        return torch.device(f'cuda:{i}')
    return torch.device('cpu')
def plot_accuracy(train_acc_all, val_acc_all):
    epochs = range(1, len(train_acc_all) + 1)
    plt.plot(epochs, train_acc_all, 'bo', label='Training acc')
    plt.plot(epochs, val_acc_all, 'b', label='Validation acc')
    plt.title('Training and validation accuracy')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.show()

batch_size, lr, num_epochs =  256, 0.05, 5

train_loss_all, train_acc_all, val_loss_all, val_acc_all = train(
    net, train_iter, val_iter, test_iter,
    num_epochs, lr, try_gpu()
)

plot_accuracy(train_acc_all, val_acc_all)
```
